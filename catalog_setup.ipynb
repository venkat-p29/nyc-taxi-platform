# Creating Catalogs

```sql
CREATE CATALOG IF NOT EXISTS dev_catalog
  COMMENT 'Development Environment - Safe to break and test things here';
```

```sql
CREATE CATALOG IF NOT EXISTS staging_catalog
  COMMENT 'Staging Environment - mirrors prod, used for validation';
```

```sql
CREATE CATALOG IF NOT EXISTS prod_catalog
  COMMENT 'Production Environment - Only promoted and tested code writes here';
```

# Creating Schemas

```sql
USE CATALOG dev_catalog;

CREATE SCHEMA IF NOT EXISTS bronze
  COMMENT 'Raw ingested data - no transformations';

CREATE SCHEMA IF NOT EXISTS silver
  COMMENT 'Clean and validated data';

CREATE SCHEMA IF NOT EXISTS gold
  COMMENT 'Business level aggregations';

CREATE SCHEMA IF NOT EXISTS audit
  COMMENT 'DQ results, pipeline logs, watermarks';
```
