```python
import pyspark.sql.functions as F
from pyspark.sql import Row
```

```python
env = dbutils.widgets.get("env")

if env not in ['dev', 'staging', 'prod']:
    raise ValueError(f"Invalid environment provided. Expected 'dev', 'staging' or 'prod'.")
```

```python
schema     = "bronze"
table      = "test"
target_tbl = f"{env}_catalog.{schema}.{table}"

print("Environment : ", env)
print("Target Table: ", target_tbl)
```

```python
spark.sql("SHOW CATALOGS").show()
```

```python
test_df = (
    spark.createDataFrame([
        Row(id=1, env=env)
    ])
)
```

```python
(
    test_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_tbl)
)

print(f"Written test_df to {target_tbl} successfully")
```
